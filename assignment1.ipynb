```sql
-- creating customers table and inserting values
REPLACE TABLE Customers1(
    CustomerID INT,
    CustomerName STRING,
    City STRING,
    AccountType STRING,
    Balance INT,
    LoanAmount INT
);
INSERT INTO Customers1 VALUES
(101,'John','Dallas','Savings',5000,10000),
(102,'Mary','Austin','Current',2500,5000),
(103,'David','Dallas','Savings',8000,0),
(104,'Emma','Houston','Savings',1500,3000),
(105,'Alex','Austin','Current',7000,12000),
(106,'Sophia','Dallas','Savings',5000,8000);
select * from Customers1;
```

```sql
--Q1 display the customername,city,balance and loanamount
SELECT CustomerName, City, Balance, LoanAmount
FROM Customers1;
```

```sql
--Q2 unique cities
SELECT DISTINCT City 
FROM Customers1;
```

```sql
--Q3 sort customers by  loanamount in descending order and limit to top 5
SELECT DISTINCT City 
FROM Customers1;

```

```sql
--Q4 show only the top 5 customers by loanamount
SELECT *
FROM Customers1
ORDER BY LoanAmount DESC;
```

```sql
--Q5 find the top 5 customers by loanamount
SELECT * FROM Customers1
ORDER BY LoanAmount DESC
LIMIT 5;
```

```sql
--Q6 find the total number of customers
SELECT COUNT(*)As Total_Customers
FROM Customers1;
```

```sql
--Q7 find the total loan amount
SELECT SUM(LoanAmount) AS Total_Loan_Amount
FROM Customers1;
```

```sql
--Q8 find the minimum balance
SELECT MIN(Balance) AS Minimum_Balance
FROM Customers1;
```

```sql
--Q9 find the maximum balance
SELECT MAX(Balance) AS Maxmimum_Balance
FROM Customers1;
```

```sql
--Q10 find the average balance
SELECT AVG(Balance) AS Average_Balance
FROM Customers1;
```

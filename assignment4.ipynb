```sql
-- Q1: Display all orders where the City is 'Dallas'.
SELECT * 
FROM customer_orders
WHERE City = 'Dallas';
```

```sql
-- Q2: Find all orders where the Price is greater than 500 AND OrderStatus is 'Delivered'.
SELECT *
FROM customer_orders
WHERE Price > 500
AND OrderStatus = 'Delivered';
```

```sql
-- Q3: Display all customers who are from Austin OR Houston.
SELECT *
FROM customer_orders
WHERE City = 'Austin'
OR City = 'Houston';
```

```sql
-- Q4: Find all orders that are NOT 'Cancelled'.
SELECT *
FROM customer_orders
WHERE NOT OrderStatus = 'Cancelled';
```

```sql
-- Q5: Display orders where the City is Dallas, Austin, or Chicago.
SELECT *
FROM customer_orders
WHERE City IN ('Dallas', 'Austin', 'Chicago');
```

```sql
-- Q6: Find products whose Price is BETWEEN 300 and 1000.
SELECT *
FROM customer_orders
WHERE Price BETWEEN 300 AND 1000;
```

```sql
-- Q7: Find all customers whose name starts with 'A' or ends with 'a'.
SELECT *
FROM customer_orders
WHERE CustomerName LIKE 'A%'
OR CustomerName LIKE '%a';
```

```sql
-- Q8: Display all records where the City IS NULL.
SELECT *
FROM customer_orders
WHERE City IS NULL;
```

```sql
-- Q9: Find the total sales amount (Quantity × Price) for each Category.
SELECT Category,
SUM(Quantity * Price) AS TotalSales
FROM customer_orders
GROUP BY Category;
```

```sql
-- Q10: Find the average product price for each SalesPerson, sorted from highest to lowest average price.
SELECT SalesPerson,
AVG(Price) AS AveragePrice
FROM customer_orders
GROUP BY SalesPerson
ORDER BY AveragePrice DESC;
```

```sql
-- Q11: Find the number of orders placed in each City but only show cities having more than 3 orders.
SELECT City,
COUNT(*) AS TotalOrders
FROM customer_orders
GROUP BY City
HAVING COUNT(*) > 3;
```

```sql
-- Q12: Display the Top 3 Products with the highest total sales amount (Quantity × Price).
SELECT Product,
SUM(Quantity * Price) AS TotalSales
FROM customer_orders
GROUP BY Product
ORDER BY TotalSales DESC
LIMIT 3;
```

```sql
-- Q1: Find total OilProduced_bbl and GasProduced_mcf
-- for each Field considering only Active wells.
SELECT
    Field,
    SUM(OilProduced_bbl) AS Total_Oil,
    SUM(GasProduced_mcf) AS Total_Gas
FROM well_production
WHERE Status = 'Active'
GROUP BY Field;
```

```sql
-- Q2: Find average OperatingCost for each Operator
-- and display from highest to lowest.
SELECT
    OperatorName,
    AVG(OperatingCost) AS Average_OperatingCost
FROM well_production
GROUP BY OperatorName
ORDER BY Average_OperatingCost DESC;
```

```sql
-- Q3: Count the number of wells in each Region.
-- Show only Regions having more than 2 wells.

SELECT
    Region,
    COUNT(*) AS Total_Wells
FROM well_production
GROUP BY Region
HAVING COUNT(*) > 2;
```

```sql
-- Q4: Display the top 3 Fields with highest total oil production.

SELECT
    Field,
    SUM(OilProduced_bbl) AS Total_Oil
FROM well_production
GROUP BY Field
ORDER BY Total_Oil DESC
LIMIT 3;
```

```sql
-- Q5: Find minimum and maximum DepthMeters
-- for each WellType excluding Abandoned wells.

SELECT
    WellType,
    MIN(DepthMeters) AS Minimum_Depth,
    MAX(DepthMeters) AS Maximum_Depth
FROM well_production
WHERE Status <> 'Abandoned'
GROUP BY WellType;
```

```sql
-- Q6: List each OperatorName with total wells
-- and average oil production.
-- Display only the top 5.

SELECT
    OperatorName,
    COUNT(*) AS Total_Wells,
    AVG(OilProduced_bbl) AS Average_OilProduced
FROM well_production
GROUP BY OperatorName
ORDER BY Average_OilProduced DESC
LIMIT 5;
```

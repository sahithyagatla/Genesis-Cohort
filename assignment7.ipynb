```sql
CREATE TABLE Laptop_Sales (
    LaptopID INT,
    Brand VARCHAR(50),
    Model VARCHAR(50),
    Price DECIMAL(10,2),
    QuantitySold INT,
    Store VARCHAR(50),
    SaleDate DATE
);
```

```sql
--1. Display laptops whose price is greater than the average laptop price

SELECT *
FROM Laptop_Sales
WHERE Price > (
    -- Find the average laptop price
    SELECT AVG(Price)
    FROM Laptop_Sales
);
```

```sql
--2. Display the most expensive laptop

SELECT *
FROM Laptop_Sales
WHERE Price = (
    -- Find the maximum laptop price
    SELECT MAX(Price)
    FROM Laptop_Sales
);
```

```sql
--3. Display the cheapest laptop

SELECT *
FROM Laptop_Sales
WHERE Price = (
    -- Find the minimum laptop price
    SELECT MIN(Price)
    FROM Laptop_Sales
);
```

```sql
--4. Display laptops sold in the same store where MacBook Air was sold

SELECT *
FROM Laptop_Sales
WHERE Store = (
    -- Find the store where MacBook Air was sold
    SELECT Store
    FROM Laptop_Sales
    WHERE Model = 'MacBook Air'
);
```

```sql
-- 5.Display laptops with quantity sold greater than the average quantity sold

SELECT *
FROM Laptop_Sales
WHERE QuantitySold > (
    -- Find the average quantity sold
    SELECT AVG(QuantitySold)
    FROM Laptop_Sales
);
```

```sql
-- 6.Display laptops sold on the latest sale date

SELECT *
FROM Laptop_Sales
WHERE SaleDate = (
    -- Find the latest sale date
    SELECT MAX(SaleDate)
    FROM Laptop_Sales
);
```

```sql
-- 7.Display laptops sold after the earliest sale date

SELECT *
FROM Laptop_Sales
WHERE SaleDate > (
    -- Find the earliest sale date
    SELECT MIN(SaleDate)
    FROM Laptop_Sales
);
```

```sql
--8. Display the brand(s) having the highest total quantity sold

SELECT Brand
FROM Laptop_Sales
GROUP BY Brand
HAVING SUM(QuantitySold) = (
    -- Find the highest total quantity sold among all brands
    SELECT MAX(TotalQty)
    FROM (
        -- Calculate total quantity sold for each brand
        SELECT SUM(QuantitySold) AS TotalQty
        FROM Laptop_Sales
        GROUP BY Brand
    ) AS BrandTotals
);
```

```sql
--9. Display laptops whose price is greater than the price of every Dell laptop

SELECT *
FROM Laptop_Sales
WHERE Price > (
    -- Find the highest Dell laptop price
    SELECT MAX(Price)
    FROM Laptop_Sales
    WHERE Brand = 'Dell'
);
```

```sql
--10. Display laptops whose price matches the price of any Apple laptop

SELECT *
FROM Laptop_Sales
WHERE Price IN (
    -- Get all Apple laptop prices
    SELECT Price
    FROM Laptop_Sales
    WHERE Brand = 'Apple'
);
```
